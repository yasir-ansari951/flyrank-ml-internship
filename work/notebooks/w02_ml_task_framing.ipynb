## My Lane as an ML Task

I selected the Content Refresh Prioritization lane.

This problem is best framed as a ranking and scoring task. The goal is not simply to classify pages as "good" or "bad", but to rank pages according to how likely they are to benefit from a content refresh.

In a real SEO workflow, content teams often manage thousands of pages. Since resources are limited, not every page can be updated immediately. A ranking system would help prioritize pages that appear to have the highest opportunity for improvement.

The output of the system would be a score or ranking that helps content teams decide which pages to review first.

## 2. Target or proxy

*What would you predict? Where does that label come from — observed outcome or a defined rule?*

## Target or Proxy

The ideal target would be the future performance improvement of a page after a content refresh. Examples could include increases in impressions, clicks, CTR, or ranking position.

However, the starter dataset does not directly contain future refresh outcomes. Because of this, a proxy target is needed.

One possible proxy is trend_direction. Pages with a declining trend may indicate content that is losing relevance, visibility, or competitiveness. These pages may be stronger candidates for content refresh activities.

This proxy is not a perfect measure of refresh opportunity, but it provides a reasonable starting point for framing the problem.

## 3. Success metric

*One metric you can defend. What number means 'good'?*

## Success Metric

A useful success metric for this problem would be Precision@K.

The purpose of the system is to recommend a limited number of pages for review. If a content team only has resources to refresh the top 50 pages, then the quality of those recommendations matters more than overall accuracy.

Precision@K measures how many of the top-ranked recommendations are actually valuable opportunities.

This metric aligns closely with the business decision because it focuses on the quality of the pages selected for action rather than evaluating every page equally.

## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*

## Why ML Beats a Fixed Rule

A fixed rule might look something like:

"Refresh every page with CTR below 0.30"

While simple, this approach ignores many other signals available in the dataset.

For example, two pages may have the same CTR but differ significantly in:

- Search volume
- Ranking position
- Competition level
- Content age
- Trend direction
- Engagement metrics

The interaction between these variables can be complex and difficult to capture with a single rule.

Machine learning can evaluate multiple signals simultaneously and identify patterns that are not obvious through manual inspection. Instead of relying on one threshold, it can produce a more nuanced ranking of refresh opportunities.

The goal is not to replace human judgment but to help content teams focus their attention where it is most likely to create value.

## Real Content Action

The output of this system would support a practical content management decision.

If the model identifies pages with a high refresh opportunity score, a content team could:

1. Review the page content.
2. Update outdated information.
3. Improve keyword coverage.
4. Add new sections or examples.
5. Improve internal linking.
6. Monitor performance after the update.

The recommendation does not automatically change content. Instead, it helps humans decide where to spend their time and resources.
