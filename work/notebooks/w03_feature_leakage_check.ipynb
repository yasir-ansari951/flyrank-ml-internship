## 1. Build the feature vector

*Code that actually builds it — engineered features, categorical handling, fills.*

```python
from google.colab import userdata
from datasets import load_dataset

HF_TOKEN = userdata.get("HF_TOKEN")

ds = load_dataset(
    "FlyRank/internship-warehouse",
    "fact_content_daily_performance",
    token=HF_TOKEN
)

print(ds)
```

```python
sample = ds["train"].select(range(10000)).to_pandas()

print(sample.shape)
sample.head(10)
```

```python
import pandas as pd

# Use a small sample
sample = ds["train"].select(range(10000)).to_pandas()

# Build Feature Vector
feature_vector = sample[[
    "gsc_impressions",
    "gsc_clicks",
    "gsc_avg_position",
    "ga4_pageviews",
    "ga4_engaged_sessions"
]].copy()

# Fill missing values
feature_vector = feature_vector.fillna(0)

# Engineered Feature: CTR
feature_vector["ctr"] = (
    feature_vector["gsc_clicks"] /
    feature_vector["gsc_impressions"].replace(0, 1)
)

# Engineered Feature: Engagement Rate
feature_vector["engagement_rate"] = (
    feature_vector["ga4_engaged_sessions"] /
    feature_vector["ga4_pageviews"].replace(0, 1)
)

# Categorical Feature
feature_vector["position_bucket"] = pd.cut(
    feature_vector["gsc_avg_position"],
    bins=[0, 5, 10, 20, 50, 100],
    labels=["Top 5", "Top 10", "Top 20", "Top 50", "50+"],
    include_lowest=True
)

print("Feature Vector Shape:", feature_vector.shape)

feature_vector.head()
```

## 2. Feature notes (meaning, missing, categorical, available-when?)

*For each feature: what it means, how missing values are handled, and whether it exists BEFORE the moment you predict.*

## Feature Notes

| Feature | Meaning | Missing Value Handling | Categorical? | Available Before Prediction? |
|---------|---------|------------------------|--------------|------------------------------|
| gsc_impressions | Total number of Google Search impressions for the page. | Missing values are filled with 0. | No | Yes. Historical Search Console data is available before making a prediction. |
| gsc_clicks | Total number of clicks from Google Search. | Missing values are filled with 0. | No | Yes. Previous click history is known before prediction. |
| gsc_avg_position | Average search ranking position of the page. | Missing values are filled with 0. | No | Yes. Current ranking information is available before the prediction time. |
| ga4_pageviews | Total page views recorded in Google Analytics 4. | Missing values are filled with 0. | No | Yes. Historical page view data is available before making the decision. |
| ga4_engaged_sessions | Number of engaged sessions from Google Analytics 4. | Missing values are filled with 0. | No | Yes. Historical engagement information is available before prediction. |
| ctr | Click-through rate calculated as clicks ÷ impressions. | Calculated after replacing zero impressions with 1 to avoid division errors. | No | Yes. It is derived from historical Search Console data. |
| engagement_rate | Engagement rate calculated as engaged sessions ÷ page views. | Calculated after replacing zero page views with 1 to avoid division errors. | No | Yes. It is derived from historical GA4 data. |
| position_bucket | Ranking category (Top 5, Top 10, Top 20, Top 50, 50+). | Created from gsc_avg_position after filling missing values. | Yes | Yes. It is derived from historical ranking information available before prediction. |

### Leakage Test

I reviewed all selected features to check for feature leakage.

- No label-derived columns were included.
- No future-window metrics were used.
- No outcome or target columns were used as input features.
- All selected features come from historical Search Console or Google Analytics data that would have been available before making a prediction.

Therefore, no obvious feature leakage was detected in this feature set.

## 4. What I excluded and why

*The list of fields you refused to use — with one line of why each.*

## What I Excluded and Why

| Excluded Field | Why I Excluded It |
|----------------|-------------------|
| client_hash_id | It is only an identifier and does not help predict content performance. Using it may cause the model to memorize specific clients instead of learning general patterns. |
| content_hash_id | It uniquely identifies a page but contains no predictive information. It could also encourage memorization rather than generalization. |
| report_date | I excluded it because the date itself is not a meaningful feature for this baseline model. It may also introduce time-related bias if not handled carefully. |
| Future performance metrics | Any metric collected after the prediction date would leak future information and make the model unrealistic. |
| Label-derived columns | Features created directly from the target or outcome were excluded because they introduce feature leakage and produce misleadingly high performance. |
| Product or internal flags | Internal flags may already encode business decisions or outcomes, so they were excluded to avoid hidden leakage. |
| Any manually created future-window features | These features would not be available when making a real prediction, so they were excluded to keep the model realistic. |

### Summary

I only kept historical Search Console and Google Analytics features that would have been available before making a prediction. All identifier fields, future information, label-derived columns, and internal product flags were excluded to prevent feature leakage and improve the model's ability to generalize.
