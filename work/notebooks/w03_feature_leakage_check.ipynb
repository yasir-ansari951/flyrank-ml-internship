## 1. Build the feature vector

*Code that actually builds it — engineered features, categorical handling, fills.*

```python
from google.colab import userdata
from datasets import load_dataset
import pandas as pd

HF_TOKEN = userdata.get("HF_TOKEN")

ds = load_dataset(
    "FlyRank/internship-warehouse",
    "fact_content_daily_performance",
    token=HF_TOKEN
)

# FIX: shuffle before sampling so this notebook's exploration reflects
# many clients, not whatever sorts first in storage.
sample = ds["train"].shuffle(seed=42).select(range(10000)).to_pandas()

print(sample.shape)
print("Unique clients in sample:", sample["client_hash_id"].nunique())
sample.head()
```

# 1. Build the feature vector

```python
feature_vector = sample[[
    "gsc_impressions",
    "gsc_clicks",
    "gsc_avg_position",
    "ga4_pageviews",
    "ga4_engaged_sessions"
]].copy()

feature_vector = feature_vector.fillna(0)

feature_vector["ctr"] = (
    feature_vector["gsc_clicks"] / feature_vector["gsc_impressions"].replace(0, 1)
)

feature_vector["engagement_rate"] = (
    feature_vector["ga4_engaged_sessions"] / feature_vector["ga4_pageviews"].replace(0, 1)
)

feature_vector["position_bucket"] = pd.cut(
    feature_vector["gsc_avg_position"],
    bins=[0, 5, 10, 20, 50, 100],
    labels=["Top 5", "Top 10", "Top 20", "Top 50", "50+"],
    include_lowest=True
)

print("Feature Vector Shape:", feature_vector.shape)
feature_vector.head()
```

## 3. The leakage hunt

*Attack your own features: label-derived columns, future windows, product flags. Show the test.*

```python
# ------------------------------------------------------------
# Leakage check, v2: available-before-prediction check PLUS an explicit
# flag for "this field is a strong candidate for defining a future label"
# ------------------------------------------------------------

leakage_check = pd.DataFrame({
    "Feature": feature_vector.columns,
    "Available_Before_Prediction": [
        "Yes", "Yes", "Yes", "Yes", "Yes", "Yes", "Yes", "Yes"
    ],
    "Candidate_Label_Defining_Field": [
        "No",   # gsc_impressions
        "YES",  # gsc_clicks  <- flagged: commonly used to define "underperforming" labels
        "YES",  # gsc_avg_position <- flagged: commonly used to define "underperforming" labels
        "No",   # ga4_pageviews
        "No",   # ga4_engaged_sessions
        "No",   # ctr (derived from clicks, inherits the flag if clicks defines the label)
        "No",   # engagement_rate
        "No",   # position_bucket (derived from avg_position, inherits the flag)
    ],
    "Note": [
        "Safe as a feature under any target definition seen so far.",
        "SAFE AS A FEATURE ONLY IF NOT ALSO USED TO BUILD THE TARGET. Must be re-checked once a target is defined.",
        "SAFE AS A FEATURE ONLY IF NOT ALSO USED TO BUILD THE TARGET. Must be re-checked once a target is defined.",
        "Safe as a feature under any target definition seen so far.",
        "Safe as a feature under any target definition seen so far.",
        "Derived from gsc_clicks — inherits its flag.",
        "Safe as a feature under any target definition seen so far.",
        "Derived from gsc_avg_position — inherits its flag.",
    ]
})

display(leakage_check)

print()
print("ACTION ITEM for whoever defines the target in a later notebook:")
print("If the target formula uses gsc_clicks and/or gsc_avg_position (or any")
print("column derived from them, like ctr or position_bucket), those columns")
print("MUST be excluded from the feature set used to train a model on that target.")
```
